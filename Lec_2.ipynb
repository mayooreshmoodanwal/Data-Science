```python
# initialising a list & adding elements to it using append , extend & insert
lst = [1, 2, 3, "hello", 'c', [2, 4], True]

print(lst)

#append
lst.append(3)
print(lst)

#append
lst.append([1, 2, 3])
print(lst)

#extend
lst.extend([1, 2, 3])
print(lst)

#insert
lst.insert(1, "Supream")
print(lst)

lst.insert(1, "is")
print(lst)

lst.insert(1, "Ayush")
print(lst)
```

```python
#removing an element from the list
lst = [1, 2, 3, "hello", 'c', [2, 4], True]

#pop()
lst.pop()
print(lst)

#remove()
lst.remove("hello")
print(lst)

#clear
lst.clear()
print(lst)
```

```python
lst = [3, 2, 1, 3, 3]

#index()
print(lst.index(2))

#count()
print(lst.count(3))


#reverse()
lst.reverse()
print(lst)

#sum(), min(), max()
print(sum(lst))
print(min(lst))
print(max(lst))


```

```python
#sort() sort just sort the list and do not return anything
#sorted() returns the sorted array and do not change the list
lst = [86, 56, 23, -67, 1, 55, 90, 23]
print(sorted(lst))
print(lst)
lst.sort()
print(lst)
```

```python
#Tuple
#functions (count, index, len, sum, max, min, sorted, return list)
tp =(1, 2, 33, -4, 56)
a = tuple(sorted(tp))
print(a)
print(tp)
print(tp.count(33))
print(tp.index(-4))
print(len(tp))
print(sum(tp))
print(max(tp))
print(min(tp))
```
