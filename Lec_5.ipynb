```python
import numpy as np
a=np.random.randint(1,90,(5,5))
print(a)
print('Column wise Mean \n',np.mean(a,0))
print('Row wise Mean \n',np.mean(a,1))
b=np.where(a%2==0,-1,a)
print('Replacing even no by -1 \n',b)
print('Column wise Sort \n',np.sort(a,0))
print('Row wise Sort \n',np.sort(a,1))
print('Column wise Sum \n',np.sum(a,0))
print('Row wise Sum \n',np.sum(a,1))
print('Count negetive',np.sum(b<0,1,0))

```

```python
st=np.random.randint(20,80,(5,5))
print(st)
de=np.random.randint(20,80,(5,5))
print(de)
rs=(st-de)
print(rs)
crs=np.where(rs<0,0,rs)
print(crs)
tot=np.sum(rs,0)
# print(tot)
cot=np.sum(crs,0)
# print(cot)
print(cot-tot)
zs=np.sum(np.where(crs==0,1,0),axis=0)
print(zs)


```
