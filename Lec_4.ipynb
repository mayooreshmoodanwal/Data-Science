```python
lst = []
for i in range(7):
    lst.append(int(input()))
a = np.array(lst)
print(a)

print('total run =',np.sum(a))
print('average run =',np.mean(a))
print('max run =',np.max(a),'on day',np.argmax(a) + 1)
print('min run =',np.min(a),'on day',np.argmin(a) + 1)
print('no of centuries =',np.sum(a>=100))
print('no of half centuries =', np.sum((a >= 50) & (a < 100)))
print('no of ducks =',np.sum(a==0))
#
```

```python
import numpy as np

n = int(input("Number of items: "))

prices = []

for i in range(n):
    prices.append(int(input()))

prices = np.array(prices)

print("Prices =", prices)

print("Total Bill    =", np.sum(prices))
print("Average Price =", np.mean(prices))
print("Minimum Price =", np.min(prices))
print("Maximum Price =", np.max(prices))
print("Std Dev       =", np.std(prices))

print("Sorted Prices =", np.sort(prices))
print("Unique Prices =", np.unique(prices))

print("Items > 100   =", prices[np.where(prices > 100)])
```
