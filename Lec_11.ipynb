```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
x=[1,2,3,4,5,6]
y=[1,4,9,16,25,36]
plt.plot(x,y)
# plt.figure(figsize=(10,10)) error aa rha hai
plt.xlabel('numbers')
plt.ylabel('squares')
plt.title('square graph')
plt.tight_layout()
plt.show()
```

```python
subjects=['maths','phy','eng','cs']
score=[70,88,76,96]
plt.figure(figsize=(3,2))
plt.bar(subjects,score, color='purple', edgecolor='black')
plt.xlabel('subjects')
plt.ylabel('score')
plt.title('bar graph')
plt.tight_layout()
plt.show()
```

```python
hours = np.random.randint(1,100,50)
marks = hours*8 + np.random.randint(100,200,50)
print(hours)
print(marks)
plt.figure(figsize=(4,3))
plt.scatter(hours,marks,s=6,c='red')## s se dot ka size define hota hai
plt.xlabel('hours')
plt.ylabel('marks')
plt.title('scatter plot')
plt.tight_layout()
plt.show()
```

```python
data= np.random.normal(60,10,500)
plt.hist(data,bins=5,edgecolor='black',)
plt.xlabel('marks')
plt.ylabel('frequency')
plt.title('histogram')
plt.tight_layout()
plt.show()
# data2= np.random.randint(1,10,50)
# plt.hist(data2)
```

```python
subjects=['maths','phy','eng','cs','dsa']
score=[70,88,76,96,41]
plt.pie(score,labels=subjects,autopct='%.2f%%',shadow=True,startangle=0,explode=(0.05,0.05,0.05,0.05,0.05))
plt.xlabel('subjects')
plt.ylabel('score')
plt.title('pie chart')
plt.tight_layout()
plt.show()
```
